```python
import pandas as pd

df = pd.read_csv("data/sampregdata.csv")
df.head()
```

```python
df.columns
```

```python
df = df.drop(columns=["Unnamed: 0"])
df.columns
```

```python
from sklearn.linear_model import LinearRegression

y = df['y']

for col in df.columns:
    if col == "y":
        continue

    X = df[[col]]

    model = LinearRegression()
    model.fit(X, y)

    score = model.score(X, y)


    print(col, score)
```

```python
X = df[["x4"]]
y = df[["y"]]

model = LinearRegression()
model.fit(X, y)

score = model.score(X, y)

print("x4", score)
```

# Model 1

```python
import joblib
joblib.dump(model, "models/model 1.pkl")
```

# Model 2

```python
X = df[["x2","x4"]]
y = df[["y"]]

model = LinearRegression()
model.fit(X, y)

score = model.score(X, y)

print("x2, x4", score)
```
